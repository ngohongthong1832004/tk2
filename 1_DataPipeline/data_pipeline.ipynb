# Yêu cầu 1: Data Pipeline
- Làm sạch và tiền xử lý dữ liệu
- Feature Engineering: TF-IDF
- Lưu dataset đã xử lý

```python
import pandas as pd
import re
import os
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer
```

## 1. Đọc dataset

```python
df = pd.read_csv('../spamDataset.csv', encoding='latin-1')
df = df[['v1', 'v2']].rename(columns={'v1': 'label', 'v2': 'text'})
df = df.dropna()
print(df.shape)
df.head()
```

## 2. Làm sạch dữ liệu
- Chuyển chữ thường
- Loại bỏ ký tự đặc biệt, số
- Loại bỏ khoảng trắng dư

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['text_clean'] = df['text'].apply(clean_text)
df['label_num'] = df['label'].map({'ham': 0, 'spam': 1})
df.head()
```

## 3. Feature Engineering — TF-IDF

```python
vectorizer = TfidfVectorizer(max_features=3000, stop_words='english')
X = vectorizer.fit_transform(df['text_clean'])
y = df['label_num'].values
print('TF-IDF shape:', X.shape)
```

## 4. Lưu dataset đã xử lý

```python
os.makedirs('processed', exist_ok=True)
df.to_csv('processed/clean_data.csv', index=False)
joblib.dump(X, 'processed/X_tfidf.pkl')
joblib.dump(y, 'processed/y.pkl')
joblib.dump(vectorizer, 'processed/tfidf_vectorizer.pkl')
print('Đã lưu dataset đã xử lý vào thư mục processed/')
```
